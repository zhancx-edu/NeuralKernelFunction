```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re
import numpy as np

import pandas as pd
import numpy as np

# ============================================================
# Compute mean and standard deviation
# ============================================================

def summarize_results(results, model_name):
    """
    results: list of tuples
        (Dataset, Total, Temporal, Spatial)
    """
    df = pd.DataFrame(
        results,
        columns=["Dataset", "Total", "Temporal", "Spatial"]
    )

    summary = (
        df.groupby("Dataset")
          .agg({
              "Total": ["mean", "std"],
              "Temporal": ["mean", "std"],
              "Spatial": ["mean", "std"],
          })
    )

    summary.columns = [
        "Total_mean", "Total_std",
        "Temporal_mean", "Temporal_std",
        "Spatial_mean", "Spatial_std",
    ]

    summary = summary.reset_index()

    rows = []

    for _, row in summary.iterrows():

        rows.append([
            row["Dataset"],
            "Temporal",
            model_name,
            row["Temporal_mean"],
            row["Temporal_std"]
        ])

        rows.append([
            row["Dataset"],
            "Spatial",
            model_name,
            row["Spatial_mean"],
            row["Spatial_std"]
        ])

        rows.append([
            row["Dataset"],
            "Total",
            model_name,
            row["Total_mean"],
            row["Total_std"]
        ])

    return rows


# ============================================================
# ETAS results (three independent runs)
# ============================================================

etas = [
    ("SCEDC_20",-5.051734447479248,2.5602879524230957,-7.612022399902344),
    ("SCEDC_25",-5.510946750640869,2.115889549255371,-7.626836776733398),
    ("SCEDC_30",-5.915470123291016,1.8469997644424438,-7.76246976852417),
    ("ComCat",-7.042874813079834,1.42739737033844,-8.4702730178833),
    ("SaltonSea",-0.19661054015159607,2.2988452911376953,-2.4954559803009033),
    ("SanJac",-4.638351917266846,1.1047427654266357,-5.7430949211120605),
    ("WHITE",-2.455518960952759,2.0299463272094727,-4.485465049743652),

    ("SCEDC_20",-5.039781093597412,2.565744638442993,-7.605525970458984),
    ("SCEDC_25",-5.5180840492248535,2.1142284870147705,-7.632312297821045),
    ("SCEDC_30",-5.920539379119873,1.8440749645233154,-7.764614105224609),
    ("ComCat",-7.042929649353027,1.4239940643310547,-8.466922760009766),
    ("SaltonSea",-0.17134174704551697,2.2803537845611572,-2.4516959190368652),
    ("SanJac",-4.639002323150635,1.106516718864441,-5.745519161224365),
    ("WHITE",-2.45760178565979,2.020827054977417,-4.478428840637207),

    ("SCEDC_20",-5.051199913024902,2.5517220497131348,-7.602923393249512),
    ("SCEDC_25",-5.506045341491699,2.1320223808288574,-7.638068199157715),
    ("SCEDC_30",-5.947542667388916,1.81911301612854,-7.766655445098877),
    ("ComCat",-7.048686504364014,1.4194324016571045,-8.468117713928223),
    ("SaltonSea",-0.18313981592655182,2.2906296253204346,-2.473769426345825),
    ("SanJac",-4.644066333770752,1.103540301322937,-5.7476067543029785),
    ("WHITE",-2.456740617752075,2.026407480239868,-4.483148097991943),
]

# ============================================================
# ST-NKF results (three independent runs)
# ============================================================

stnkf = [
    ("SCEDC_20",-5.07567024230957,2.5326812267303467,-7.60835075378418),
    ("SCEDC_25",-5.519142150878906,2.105340003967285,-7.62448263168335),
    ("SCEDC_30",-5.991769313812256,1.7406789064407349,-7.732447624206543),
    ("ComCat",-7.061898708343506,1.4242260456085205,-8.486124992370605),
    ("SaltonSea",-0.16384480893611908,2.290472984313965,-2.454317569732666),
    ("SanJac",-4.482001304626465,1.1184521913528442,-5.600453853607178),
    ("WHITE",-2.3940441608428955,2.025203227996826,-4.419247627258301),

    ("SCEDC_20",-5.082316875457764,2.54353404045105,-7.625850200653076),
    ("SCEDC_25",-5.539050579071045,2.103717565536499,-7.642767906188965),
    ("SCEDC_30",-5.993575572967529,1.7514182329177856,-7.744994163513184),
    ("ComCat",-7.059010028839111,1.4232944250106812,-8.482303619384766),
    ("SaltonSea",-0.16416627168655396,2.318432092666626,-2.4825990200042725),
    ("SanJac",-4.4921722412109375,1.1076663732528687,-5.5998382568359375),
    ("WHITE",-2.388584613800049,2.0256471633911133,-4.414231777191162),

    ("SCEDC_20",-5.062490463256836,2.5517590045928955,-7.614248752593994),
    ("SCEDC_25",-5.55588960647583,2.087313175201416,-7.643203258514404),
    ("SCEDC_30",-5.927818298339844,1.8030840158462524,-7.730902194976807),
    ("ComCat",-7.058920860290527,1.4143037796020508,-8.473224639892578),
    ("SaltonSea",-0.19460539519786835,2.3002548217773438,-2.4948604106903076),
    ("SanJac",-4.483128547668457,1.1122349500656128,-5.595363616943359),
    ("WHITE",-2.3976755142211914,2.028867721557617,-4.426543712615967),
]

# ============================================================
# Build plotting dataframe
# ============================================================

rows = []
rows.extend(summarize_results(stnkf, "ST-NKF"))
rows.extend(summarize_results(etas, "ETAS"))

df = pd.DataFrame(
    rows,
    columns=[
        "Dataset",
        "Metric",
        "Model",
        "LogLikelihood",
        "StdDev",
    ],
)

print(df)


sns.set_theme(style="whitegrid", context="paper", font="serif")

fig, axes = plt.subplots(1, 3, figsize=(14, 4.5), sharey=False)

metrics = ['Temporal', 'Spatial', 'Total']
titles = [
    '(a) Marginal Temporal Log-Likelihood',
    '(b) Conditional Spatial Log-Likelihood',
    '(c) Total Log-Likelihood'
]

palette = sns.color_palette("colorblind", 2)
model_colors = {"ST-NKF": palette[0], "ETAS": palette[1]}


for ax, metric, title in zip(axes, metrics, titles):
    metric_df = df[df['Metric'] == metric]
    
    datasets = metric_df['Dataset'].unique()
    x = np.arange(len(datasets))
    width = 0.35 
    
    st_nkf_data = metric_df[metric_df['Model'] == 'ST-NKF']
    etas_data = metric_df[metric_df['Model'] == 'ETAS']


    bars1 = ax.bar(x - width/2, st_nkf_data['LogLikelihood'], width, 
                   label='ST-NKF', color=model_colors['ST-NKF'],
                   yerr=st_nkf_data['StdDev'], capsize=4, ecolor='black', 
                   edgecolor='black', linewidth=0.5)


    bars2 = ax.bar(
        x + width/2,
        etas_data["LogLikelihood"],
        width,
        label="ETAS",
        color=model_colors["ETAS"],
        yerr=etas_data["StdDev"],     
        capsize=4,                     
        ecolor="black",               
        edgecolor="black",
        linewidth=0.5,
    )

    ax.set_ylabel('Log-Likelihood', fontsize=10)
    ax.set_title(title, fontsize=12, pad=10)
    ax.set_xticks(x)
    ax.set_xticklabels(datasets, rotation=45, ha='right', fontsize=9)
    ax.tick_params(axis='y', labelsize=9)
    ax.axhline(0, color='black', linewidth=0.7)

    sns.despine(ax=ax, left=False, bottom=False) 
    ax.grid(axis='y', linestyle='--', alpha=0.6) 
    ax.grid(axis='x', linestyle='', alpha=0)


handles, labels = ax.get_legend_handles_labels()

fig.legend(handles, labels, loc='upper center', bbox_to_anchor=(0.5, 0), ncol=2, frameon=False, fontsize=10)

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.savefig('log_likelihood_comparison_sci_style.png', dpi=300, bbox_inches='tight')
plt.savefig('log_likelihood_comparison_sci_style.pdf', bbox_inches='tight') 

```
