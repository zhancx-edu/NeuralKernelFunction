```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import os
import math
import pandas as pd
from matplotlib import colors

def plot_scatter(x_arr, y_arr, c_arr, time_arr, training_num=None, validation_num=None, label=None, y_label=None):
    """
    绘制散点图，底部x轴为数据索引，顶部x轴为每两年一次的时间年份。

    Args:
        x_arr (array-like): x坐标数组 (通常是数据点的索引)。
        y_arr (array-like): y坐标数组。
        c_arr (array-like): 用于颜色映射的值数组。
        time_arr (array-like): 时间字符串数组 (格式: "YYYY-MM-DD HH:MM:SS.sss")。
        training_num (int, optional): 训练集结束的索引。
        validation_num (int, optional): 验证集的数据点数量。
        label (str, optional): 用于生成文件名的标签。
        y_label (str, optional): y轴的标签。
    """
    cmap = plt.get_cmap('coolwarm')
    vmax = np.percentile(np.abs(c_arr), 90)
    norm = colors.Normalize(vmin=-vmax, vmax=vmax)

    fig, ax = plt.subplots(figsize=(12, 6), dpi=300)

    # --- 1. 创建共享Y轴的第二个X轴（位于顶部）---
    ax2 = ax.twiny()

    # 散点图部分
    etas_sc = ax.scatter(x_arr, y_arr, cmap=cmap, norm=norm, c=c_arr, s=1)

    # --- 2. 改进年份刻度逻辑：每隔两年显示一次 ---
    all_year_tick_positions = []
    all_year_labels = []
    seen_years = set()
    last_labeled_year = -9999  # 初始化一个不可能的年份


    for i, time_str in enumerate(time_arr):
        year_str = time_str.split('-')[0]
        if year_str not in seen_years:
            seen_years.add(year_str)
            year_int = int(year_str)
            
            # 1. 为每个新年份记录刻度位置
            all_year_tick_positions.append(x_arr[i])
            
            # 2. 判断是否为该刻度添加标签
            if last_labeled_year == -9999 or (year_int - last_labeled_year) >= 2:
                all_year_labels.append(year_str)  # 添加年份标签
                last_labeled_year = year_int
            else:
                all_year_labels.append("")  # 为不显示的年份添加空标签
    # ---------------------------------------------------

    if training_num is not None and validation_num is not None:
        ax.axvline(x=training_num, color='grey', linestyle='--', label='train / val')
        ax.axvline(x=training_num + validation_num, color='grey', linestyle='--')
        
        text_props = dict(fontsize=10, fontfamily='serif', bbox=dict(facecolor='white', edgecolor='black', boxstyle='square,pad=0.3', alpha=0.9), ha='center', va='top')
        ax.text(training_num / 2, ax.get_ylim()[1] * 0.95, 'Training Set', **text_props)
        ax.text(training_num + validation_num / 2, ax.get_ylim()[1] * 0.95, 'Validation Set', **text_props)
        ax.text(training_num + validation_num + (ax.get_xlim()[1] - (training_num + validation_num)) / 2, ax.get_ylim()[1] * 0.95, 'Test Set', **text_props)

    # --- 3. 设置底部主轴 (ax) ---
    ax.set_xlabel("Index of Earthquakes", fontsize=14)
    ax.set_ylabel(f"{y_label}", fontsize=14)
    ax.tick_params(axis='x', labelsize=12)
    ax.tick_params(axis='y', labelsize=12)

    # --- 4. 设置顶部次轴 (ax2) ---

    # 设置顶部次轴 (ax2)
    ax2.set_xlabel("Time (Year)", fontsize=14)
    ax2.set_xlim(ax.get_xlim())
    
    # 应用完整的刻度位置和包含空字符串的标签列表
    ax2.set_xticks(all_year_tick_positions)
    ax2.set_xticklabels(all_year_labels, rotation=45, fontsize=12)

    # 美化边框
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_linewidth(1.5)
    ax.spines['left'].set_linewidth(1.5)
    ax.tick_params(width=1.5)
    
    ax2.spines['bottom'].set_visible(False)
    ax2.spines['right'].set_visible(False)
    ax2.spines['left'].set_visible(False)
    ax2.spines['top'].set_linewidth(1.5)
    ax2.tick_params(axis='x', width=1.5)

    ax.grid(True, linestyle="--", alpha=0.5)

    # --- 修改 colorbar 的添加方式 ---
    # 为 colorbar 创建一个新的 axes，位于主图的右边
    cbar_ax = fig.add_axes([0.98, 0.15, 0.01, 0.65])  # [left, bottom, width, height]
    cbar = fig.colorbar(etas_sc, cax=cbar_ax, orientation='vertical', extend='both')
    # ----------------------------------

    if "spatial" in label:
        cbar.set_label(r'$\mathcal{L}_{Spatial, ST-NKF} - \mathcal{L}_{Spatial, ETAS}$', fontsize=13)
    elif "temporal" in label:
        cbar.set_label(r'$\mathcal{L}_{Temporal, ST-NKF} - \mathcal{L}_{Temporal, ETAS}$', fontsize=13)
    cbar.ax.tick_params(labelsize = 11)


    plt.tight_layout()

    if not os.path.exists('fig'):
        os.makedirs('fig')
        
    plt.savefig(f"./fig/scatter-{label}-{y_label}.png", dpi=500, bbox_inches='tight')
    plt.close()
    print(f"Plot saved to ./fig/{label}-{y_label}.png")


for label in ['ComCat', 'SaltonSea', 'SanJac', 'WHITE', 'SCEDC_20', 'SCEDC_25', 'SCEDC_30']:
    
    data_etas = pd.read_csv(f'./csv/{label}_empirical_empirical_empirical.csv')

    index_etas = data_etas['step'].isin(['train', 'val', 'test'])

    intlam_etas = data_etas['intlam'][index_etas].values
    ll_t_etas = data_etas['logli_t'][index_etas].values
    ll_s_etas = data_etas['logli_s'][index_etas].values

    time_etas = data_etas['time'][index_etas].values
    magnitude_etas = data_etas['magnitude'][index_etas].values
    longitude_etas = data_etas['longitude'][index_etas].values
    latitude_etas = data_etas['latitude'][index_etas].values


    data_nkf = pd.read_csv(f'./csv/nkf/{label}_neural_neural_neural.csv')

    index_nkf = data_nkf['step'].isin(['train', 'val', 'test'])

    intlam_nkf = data_nkf['intlam'][index_nkf].values
    ll_t_nkf = data_nkf['logli_t'][index_nkf].values
    ll_s_nkf = data_nkf['logli_s'][index_nkf].values

    time_nkf = data_nkf['time'][index_nkf].values
    magnitude_nkf = data_nkf['magnitude'][index_nkf].values
    longitude_nkf = data_nkf['longitude'][index_nkf].values
    latitude_nkf = data_nkf['latitude'][index_nkf].values

    train_num = np.sum(data_etas['step'].isin(['train']))
    val_num = np.sum(data_etas['step'].isin(['val']))
    test_num = np.sum(data_etas['step'].isin(['test']))
    print(train_num, val_num, test_num, train_num + val_num + test_num)

    train_num = np.sum(data_etas['step'].isin(['train']))
    val_num = np.sum(data_etas['step'].isin(['val']))

    plot_scatter(np.arange(1, magnitude_etas.shape[0] + 1), magnitude_etas, ll_t_nkf - ll_t_etas, time_etas, train_num, val_num, label = f"{label}_temporal", y_label = "Magnitude")
```
